```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END 

class State(TypedDict):
    text: str
    word_count: int 

def clean(state: State) -> State:
    return {
        "text": state["text"].strip().lower()
    }
    
def word_count(state: State) -> State:
    return {
        "word_count": len(state["text"].split())
    }
    
builder = StateGraph(State)
builder.add_node("clean", clean)
builder.add_node("word_count", word_count)

builder.add_edge(START, "clean")
builder.add_edge("clean", "word_count")
builder.add_edge("word_count", END)

graph = builder.compile()
```

```python
result = graph.invoke({"text": "  Hello World! This is a test.  "})
print(result)  # Output: {'text': 'hello world! this is a test.',
```
