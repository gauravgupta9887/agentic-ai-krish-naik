```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class State(TypedDict):
    text: str
    word_count: int

def clean(state): return {"text": state["text"].strip().lower()}
def count(state): return {"word_count": len(state["text"].split())}

g = StateGraph(State)
g.add_node("clean", clean); g.add_node("count", count)
g.add_edge(START, "clean"); g.add_edge("clean", "count"); g.add_edge("count", END)
graph = g.compile()
inp = {"text": "  Hello LangGraph World  "}

print("stream_mode='updates' (what each node returned):")
for chunk in graph.stream(inp, stream_mode="updates"):
    print("  ", chunk)
print("stream_mode='values' (full state after each step):")
for chunk in graph.stream(inp, stream_mode="values"):
    print("  ", chunk)
```
