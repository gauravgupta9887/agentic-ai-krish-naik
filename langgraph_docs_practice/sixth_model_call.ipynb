```python
from typing import TypedDict
from langchain_core.language_models.fake_chat_models import FakeListChatModel
from langgraph.graph import StateGraph, START, END
```

```python
class State(TypedDict):
    ticket: str
    reply: str

def make_graph(model):
    def draft(state: State) -> dict:
        resp = model.invoke(f"Write a short reply to this ticket: {state['ticket']}")
        return {"reply": resp.content}
    g = StateGraph(State)
    g.add_node("draft", draft)
    g.add_edge(START, "draft"); g.add_edge("draft", END)
    return g.compile()
```

```python
fake = FakeListChatModel(responses=["Sorry about the double charge. We have refunded it."])
print(make_graph(fake).invoke({"ticket": "Charged twice!", "reply": ""}))
```
