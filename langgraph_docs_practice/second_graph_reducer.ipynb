```python
import operator
from typing import TypedDict, Annotated
from langgraph.graph import StateGraph, START, END 

class NoReducer(TypedDict):
    log: list[str]
    
class WithReducer(TypedDict):
    log: Annotated[list[str], operator.add]
    
def make(schema):
    def a(state): return {"log": ["a run"]}
    def b(state): return {"log": ["b run"]}
    g = StateGraph(schema)
    g.add_node("a", a)
    g.add_node("b", b) 
    g.add_edge(START, "a")
    g.add_edge("a", "b")
    g.add_edge("b", END)
    return g.compile()
```

```python
print(" no reducer:", make(NoReducer).invoke({"log":[]}))
print(" with reducer:", make(WithReducer).invoke({"log":[]}))
```
