```python
from typing import TypedDict, Annotated, List
from langgraph.graph import StateGraph, START, END
import operator as op
```

```python
class State(TypedDict):
    logs: Annotated[List[str], op.add]
    logs_not_reduced: List[str]
```

```python
def writer_node_1(state:State)->dict :
    return {"logs":["Logs by node writer 1"], "logs_not_reduced":["Logs by node writer 1"]}

def writer_node_2(state:State)->dict:
    return {"logs":["Logs by node writer 2"], "logs_not_reduced":["Logs by node writer 2"]}
        
    
def writer_node_3(state:State)->dict:
    return {"logs":["Logs by node writer 3"], "logs_not_reduced":["Logs by node writer 3"]}
    
```

```python
builder = StateGraph(State)

builder.add_node("writer_node_1", writer_node_1)
builder.add_node("writer_node_2", writer_node_2)
builder.add_node("writer_node_3", writer_node_3)
builder.add_edge(START, "writer_node_1")
builder.add_edge("writer_node_1", "writer_node_2")
builder.add_edge("writer_node_2", "writer_node_3")
builder.add_edge("writer_node_3", END)

graph = builder.compile()
```

```python
for chunk in graph.stream({"logs":[],"logs_not_reduced":[]},stream_mode="values"):
    print(chunk)
```
